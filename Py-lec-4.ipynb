TOPIC - LISTS AND TUPLES IN PYTHON

- list are denoted by '[ ]' brackets.
- list are ordered, mutable and allow duplicate values.
- list are built-in data structures
- list stores elements of different data types like integer , float , string, etc.

```python
# creating a list
marks = [22,23,56,45,55]
print(marks)
print(type(marks))
```

```python
student = ["karan" , 95.6 ,15 , "Azamgarh"]
print(student)
student[0] = "arjun"
print(student)
```

List Methods

```python
list = [2,3,5,7,1,1]
print(list)
print(type(list))

#append method
#list.append('Ayushi')
#print(list)

#sort method
# list.sort(reverse=True) for printing in descending order
list.sort()
print(list)

#count method
list.count(1)

#reverses the list
list1 = [22,33,76,12,21]
list1.reverse()
print(list1)

#copy method 
list1.copy()
print(list1)

#extend method
list1.extend([24,67,56,43])
print(list1)

#index method 
list1.insert(2,4)
print(list1)

#pop method 
list1.pop(2)
print(list1)

#remove method
list1.remove(33)
print(list1)

#clear method
list1.clear()
print(list1)
```

TUPLES IN PYTHON

- A built-in data type, creates immutable sequences of values 
- Tuples are used to store multiple items in a single variable
- It is denoted by '()' brackets .
- Tuples are similar to lists, but unlike lists they cannot be changed after their creation. 

```python
tup = (37,87,56,34,21)
print(type(tup))
print(tup[4])
```

```python
#Tuple Packing
fruits = ("apple","banana","cherry","guava")
print(fruits)
```

```python
#Tuple Unpacking
fruits= ("apple","banana","cherry","mango","Papaya")
(green,yellow,red,blue,orange,) = fruits

print(green)
print(yellow)
print(red)
print(blue)
print(orange)
```

```python
# if the number of elements doesnot match with the number of variables, use asterik (*)
numbers= (11,12,13,14,15)

first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
#Updating Tuples through converting into list first and then convert it into  the tuple
fruits_list = list(fruits)
fruits_list[2]="Pomegranate"
fruits_list[3]="Grapes"

fruits= tuple(fruits_list)
print(fruits)
```

```python
tup1 = (1,2,3)
tup2 = (4,5,6)

#Tuple Concatenation
tup3 = tup1 + tup2
print(tup3)

#Tuple Repetition
repeat = tup1 * 2
print(repeat)
```

```python
# Loop through a Tuples 
apps = ("Myntra", "Flipkart","Snapchat","Whatsapp")
#for x in apps:
#    print(x)

#for i in range (len(apps)):
#   print(apps[i])

i = 0
while i<len(apps):
    print(apps[i])
    i = i+1
```

```python
# Tuple Methods
var = (1,2,1,3,5,3,4,2,4)
print(var.count(5)) # count = helps to counts teh total occurences

print(var.index(2)) # index = returns index of first occurrences
```
