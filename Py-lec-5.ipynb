Dictionaries in Python
- a built -in data structures used to store the data values in key - value pairs.
- dictionaries is an ordered, mutable and do not allow duplicates 
- dictionaries are denoted by '{}' 
- dictionaries cannot have the two items with the same key name.

```python
info ={
    "name" : "Ayushi",
    "age": 23 ,
    "learning" : "coding" ,
    "is_kind" : True ,
    "Percentage": 87.5 
}

print(info)
```

Nested Dictionary
- a nested dictionary is a dictionary under another dictionary.
- It allows hierarchical data storage, making it useful for complex data structures.

```python
#nested dictionary examples
students = {
    "student1" : {
        'name': 'Ayushi',
        'age' : 22,
        'marks' : {'math' : 90, 'science' : 95}
    },
    "student2" : {
        'name' : 'Kartik',
        'age' : 21,
        'marks' : {'math' : 89, 'science' : 92}
    }
}
print(students)
print(students['student1']['marks'])
```

```python
# Modifying nested dictionary
students['student1']['name']= 'Snehil'
print(students)
print(students['student1'])
```

Sets In Python

- A set is an unorderd collection of unique elements.
- It is defined using curly braces '{}' or the set() instructor.
    - Unordered
    - unique elements
    - mutable
    - immutable elements

```python
set5 = {1,2,3,4,5}
print(set5)
print(type(set5))
```

```python
del set
```

```python
my_set = set([1,2,3,1,3,5])
my_set
```

```python
# Set Methods
# Adding and Removing elements
s = {1,2,3}
s.add(4)
s.update([5,6])
s.remove(1)
s.discard(10)
s.pop()
s.clear()
s
```

```python
# Set Operations
a = {1,2,3,4}
b = {3,4,5,6}

print(a|b)  #union
print(a&b)  #intersection
print(a-b)  #difference
print(a^b)  #symmetric Difference
```

```python
# Set Comparison Methods
a = {1,2}
b = {1,2,3,4}

print(a.issubset(b))
print(b.issuperset(a))
print(a.isdisjoint({1,2}))
```

```python
# Copy of a set
original  = {1,2,3}
copy_set = original.copy()
print(copy_set)
```
