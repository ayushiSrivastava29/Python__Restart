Python OOPs Concept

- OOP means Object-Oriented Programming.
- It is a way of writing programs where we think of everything as objects.
- OOP makes code organized, reusable, and easy to maintain.

OOPs consists of -- class and Objects ....
- Object → like a real-world thing (a car, a student, a dog).
- Class → a blueprint or design to create objects (e.g., a "Car" class can create many car objects).
- Main idea: Instead of just writing functions and code, we organize code into objects that have:
    - Properties (data → like color, name, age).
    - Methods (actions → like drive, bark, study).

```python
class car:
    def __init__ (self,name,model,year):
        self.name = name
        self.model = model
        self.year = year
    
c1 = car("BMW","X5",2020)
print(c1.name)
print(c1.model)
print(c1.year)

c2 = car("Audi","A6",2021)
print(c2.name)
print(c2.model) 
print(c2.year)
```

What is a Method? 

A method is simply a function inside a class that defines what an object can do (its behavior).

👉 Example: If a Dog is a class:

Attributes = name, breed, color

Methods = bark(), eat(), sleep()

So, methods are actions of the object.

```python
class Student: 
    def __init__ (self,name,age,rollno):
        self.name = name
        self.age = age
        self.rollno = rollno
    
    def welcome(self):
        print("Welcome to the class",self.name,"having roll number",self.rollno)

    def display(self):
        print(self.name)
        print(self.age)
        print(self.rollno)

    def get_marks(self,marks):
        self.marks = marks
        print("Marks of",self.name,"is",self.marks)

s1 = Student("Ayushi",20,101)
s1.display()
s1.welcome()
s1.get_marks(95)
```
